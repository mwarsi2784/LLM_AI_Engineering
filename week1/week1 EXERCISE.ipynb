To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
base_url_gpt = "https://api.openai.com/v1"
MODEL_LLAMA = 'llama3.2:latest'
base_url_llama = "http://localhost:11434/v1"
```

```python
# set up environment
load_dotenv(override=True)
api_key_gpt =os.getenv("OPENAI_API_KEY")
api_key_llama = ""  #no need to enter key
```

```python
openai = OpenAI(base_url=base_url_gpt, api_key=api_key_gpt)
llama = OpenAI(base_url=base_url_llama, api_key=api_key_llama)
```

```python
# here is the question; type over this to ask something new

question = """
Explain polymorphism in OOP in short paragraph.
"""
```

```python
system_prompt = "You are a programming teacher here. You have very good sense of humour as well"
```

```python
message = [
    {"role":"system", "content":system_prompt},
    {"role":"user","content":question}
]
```

```python
# Get gpt-4o-mini to answer, with streaming
response_gpt = openai.chat.completions.create(
    model=MODEL_GPT,
    messages= message,
    stream=True
    )
print(response_gpt)
```

```python
# Get Llama 3.2 to answer
response_llama = llama.chat.completions.create(
    model=MODEL_LLAMA,
    messages=message,
    stream=True
)
print(response_llama)
```

```python
output_gpt=""
display_handle_gpt = display(Markdown(""),display_id=True)
for chunk in response_gpt:
    output_gpt+= chunk.choices[0].delta.content or ""
    update_display(Markdown(output_gpt),display_id=display_handle_gpt.display_id)
    
```

```python
output_llama=""
display_handle_llama = display(Markdown(""),display_id=True)
for chunk in response_llama:
    output_llama+= chunk.choices[0].delta.content 
    update_display(Markdown(output_llama),display_id=display_handle_llama.display_id)
```
